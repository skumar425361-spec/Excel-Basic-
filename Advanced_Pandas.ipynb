##**Q2 Sales Filtering Analysis**

Filter the dataset to display:

##**Q3 Customer Purchase Analysis**

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.

```python
#Identify the top 5 customers based on total sales.
customer_sales = df.groupby('Customer')['Sales'].sum()
topcustomers = customer_sales.sort_values(ascending=False).head(5)

print("Top 5 Customers Based on Total Sales:")
print(topcustomers)

```

##**Q4 Regional Sales Performance**

Calculate the total sales generated by each region.

Then identify:

* Which region generated the highest revenue?

```python
regional_sales = df.groupby('Region')['Sales'].sum()
print("\n --Total Sales by Region:--")
print(regional_sales)

# Find the region with the highest revenue
highest_region = regional_sales.idxmax()
highest_revenue = regional_sales.max()

print("\nRegion with the Highest Revenue:")
print(f"{highest_region} : {highest_revenue}")
```

##**Q5 Product Performance Analysis**

* Calculate the total quantity sold for each product.

* Which product appears to be the most popular?

```python
# Calculate the total quantity sold for each product.
product_qty = df.groupby('Product')['Quantity'].sum().reset_index()

product_qty_sorted = product_qty.sort_values(by='Quantity', ascending=False)

print("---Total Quantity Sold by Product---")
print(product_qty_sorted.to_string(index=False))
```

```python
# Which product appears to be the most popular?
most_popular = product_qty_sorted.iloc[0]
most_popular_product = most_popular['Product']
total_units = most_popular['Quantity']

print(f"\n★ Most Popular Product: '{most_popular_product}' with a total of {total_units} units sold.")
```

##**Q6 Multi-Level Grouping**

* Group the dataset by: Region and Category

* Calculate the total sales for each combination.

* What insights can you derive from this analysis?

```python
# Group the dataset by: Region and Category
# Calculate the total sales for each combination.

multi_level_sales = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()
print("--- Total Sales by Region and Category Combination ---")
print(multi_level_sales.to_string(index=False))
```

##**Insights**

* Helps compare category performance across regions.
* Shows which region generates the highest sales.
* Highlights low-performing categories that may need improvement.
* Supports better marketing and inventory planning.

##**Q7 Pivot Table Creation**

Create a pivot table showing:

* Rows → Region

* Columns → Product

* Values → Total Sales

Which region sells the most laptops?


```python
#Pivot Table

pivot_sales = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print("--- Sales Pivot Table (Region vs Product) ---")
print(pivot_sales)
```

```python
#Which region sells the most laptops?
if 'Laptop' in pivot_sales.columns:
    laptop_region = pivot_sales['Laptop'].idxmax()
    laptop_sales = pivot_sales['Laptop'].max()
    print(f"\n sells the most Laptops is: '{laptop_region}' with a total revenue of ${laptop_sales:,}")
else:
    print("\n 'Laptop' column not found in your dataset. Please check your specific product")
```

##**Q8 Profitability Analysis**

* Calculate the average profit for each product category.

* Which category is most profitable?

```python
# Calculate the average profit for each product category.
category_profit = df.groupby('Category')['Profit'].mean().reset_index()
category_profit_sorted = category_profit.sort_values(by='Profit', ascending=False)

print("---Average Profit by Product Category---")
print(category_profit_sorted.to_string(index=False))
```

```python
#  Which category is most profitable?

most_profitable_row = category_profit_sorted.iloc[0]
most_profitable_cat = most_profitable_row['Category']
avg_profit_val = most_profitable_row['Profit']

print(f"\n Most Profitable Category: '{most_profitable_cat}' with an average profit of ${avg_profit_val:,.2f} per order.")
```

##**Q9 Discount Impact Study**

Analyze the relationship between discount and profit.

Tasks:

* Calculate the average profit for each discount level

* Identify whether higher discounts lead to lower profit

```python

discount_impact = df.groupby('Discount')['Profit'].mean().reset_index()

discount_impact_sorted = discount_impact.sort_values(by='Discount')

print("--- Average Profit by Discount Level ---")
print(discount_impact_sorted.to_string(index=False))
```

##**Q10 End-to-End Data Wrangling Task**

Perform the following workflow:

1. Filter the dataset where Sales > 3000

 2. Group data by Region and Product

3. Calculate total sales

4. Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# 1. Filter the dataset where Sales > 3000
high_value_orders = df[df['Sales'] > 3000]
print("--- High-Value Orders ---")
print(high_value_orders)
```

```python
# 2 & 3. Group data by Region and Product, then calculate total sales
grouped_sales = high_value_orders.groupby(['Region', 'Product'])['Sales'].sum().reset_index()
discount_impact_sorted = discount_impact.sort_values(by='Discount')

print("--- Average Profit by Discount Level ---")
print(discount_impact_sorted.to_string(index=False))
```

```python
#4 . Reshape the result into a final pivot table representation
# Rows -> Region, Columns -> Product_Name, Values -> Sales
final_wrangling_pivot = grouped_sales.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    fill_value=0)

print("--- Final Reshaped High-Value Sales Pivot Table ---")
print(final_wrangling_pivot)
```
