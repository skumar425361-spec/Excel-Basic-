###7. Group the dataset by Product Category and calculate the total revenue generated by each category

###8. Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue and the bottom 3 categories with the lowest revenue.

####**Answer**


```python
import pandas as pd
import numpy as np

df = pd.read_excel("/content/retail_dataset .xlsx")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

category_revenue = df.groupby("Product Category")["Revenue"].sum()

top_3 = category_revenue.sort_values(ascending=False).head(3)

bottom_3 = category_revenue.sort_values(ascending=True).head(3)

print("Top 3 Product Categories by Revenue:")
print(top_3)

print("\nBottom 3 Product Categories by Revenue:")
print(bottom_3)
```

###9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue and the region with the lowest revenue.

####**Answer**

```python
import pandas as pd
import numpy as np

df = pd.read_excel("/content/retail_dataset .xlsx")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

region_revenue = df.groupby("Region")["Revenue"].sum()

highest_region = region_revenue.idxmax()
highest_value = region_revenue.max()

lowest_region = region_revenue.idxmin()
lowest_value = region_revenue.min()

print("Total Revenue by Region:")
print(region_revenue)

print("\nHighest Revenue Region:")
print(highest_region, "-", highest_value)

print("\nLowest Revenue Region:")
print(lowest_region, "-", lowest_value)
```

###10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column using NumPy

####**Answer**

```python
import pandas as pd
import numpy as np

df = pd.read_excel("/content/retail_dataset .xlsx")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print("Total Monthly Revenue:")
print(monthly_revenue)

mean_revenue = np.mean(df["Revenue"])
median_revenue = np.median(df["Revenue"])
std_revenue = np.std(df["Revenue"])

print("\nRevenue Statistics:")
print("Mean:", mean_revenue)
print("Median:", median_revenue)
print("Standard Deviation:", std_revenue)
```
