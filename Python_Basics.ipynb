##2. Explain the difference between List and Tuple in Python.

####**Answer:** In Python, List and Tuple are both used to store multiple items in a single variable, but they have important differences.

###**1. Mutability**
* List: Mutable (can be changed after creation)


```python
my_list = [1, 2, 3]
my_list[0] = 10   #allowed

```

* Tuple: Immutable (cannot be changed after creation)

```python
my_tuple = (1, 2, 3)
my_tuple[0] = 10  #not allowed
```

####**2. Syntax**
* List uses square brackets: []
* Tuple uses parentheses: ()

####**3. Performance**
* List: Slower because it is mutable.
* Tuple: Faster because it is immutable and more memory efficient.

####**4. Methods Available**
* List: Has many methods like append(), remove(), pop()
* Tuple: Has limited methods like count() and index()


##3. What is a function in Python? Why are functions useful?

Answer: A function in Python is a block of code that is used to perform a specific task. It runs only when it is called. Functions help to organize code into small and reusable parts.



```python
#Example:

def greet():
    print("Hello, welcome!")

greet()
```

##Why are functions useful?
####**a. Code Reusability**
Functions allow us to write code once and use it multiple times, which saves time and effort.
####**b. Improves Code Organization**
They divide large programs into smaller parts, making the program easy to understand.
####**c. Easy to Debug and Maintain**
If any error occurs, we can fix it inside the function without changing the whole program.
####**d. Improves Readability**
Functions make the code clean and structured, so it is easier to read and manage.

##5. Write a Python program to check whether a number is even or odd using conditional statements.

####**Answer:**

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is an Even number")
else:
    print(num, "is an Odd number")

```

##8. Write a Python program to remove duplicate values from a list using a set.

####**Answer:**


```python
numbers = [1, 2, 3, 2, 4, 5, 3, 6]
unique_numbers = list(set(numbers))

print("List after removing duplicates:", unique_numbers)
```

##9. Write a function that returns the square of a number.

##10. Write a Python program to count how many times a number appears in a list. Example List [2,3,4,2,5,2]

####**Answer:**


```python
numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print(num, "appears", count, "times in the list")
```
